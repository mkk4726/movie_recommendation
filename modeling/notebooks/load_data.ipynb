```python
file_path = '/Users/visuworks/Desktop/movie_recommendation/data_scraping/data/custom_movie_rating.txt'

rows = []

with open(file_path, 'r', encoding='utf-8') as f:
    for line_num, line in enumerate(f, 1):
        if line.strip():
            row = line.strip()
            splitted_row = row.split('/')
            
            middle = ["|".join(splitted_row[2:-1])]

            rows.append(splitted_row[:2] + middle + splitted_row[-1:])
```

```python
# 각 row 길이에 대해, 4~9개의 row 중 하나씩만 출력
targets = [4, 5, 6, 7, 9]
found = {k: False for k in targets}

for row in rows:
    l = len(row)
    if l in targets and not found[l]:
        print(f"요소 개수 {l}개인 row 예시: {row}")
        found[l] = True
    if all(found.values()):
        break
# 찾지 못한 길이도 알기 쉽게 출력
for k in targets:
    if not found[k]:
        print(f"(요소 개수 {k}개인 row는 존재하지 않음)")
```
