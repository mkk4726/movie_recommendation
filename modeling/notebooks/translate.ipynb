```python
from transformers import MarianMTModel, MarianTokenizer

model_name = "Helsinki-NLP/opus-mt-ko-en"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

def translate_ko_to_en(text: str):
    encoded = tokenizer(text, return_tensors="pt", padding=True, truncation=True)
    translated = model.generate(**encoded)
    return tokenizer.decode(translated[0], skip_special_tokens=True)

print(translate_ko_to_en("Middle-aged or old-aged men's documentary movie"))
```

```python
from langdetect import detect

def detect_language(text: str) -> str:
    try:
        lang = detect(text)
        return lang   # 'ko', 'en', 'ja', ...
    except:
        return "unknown"

# 사용 예
print(detect_language("밝은 여름 느낌의 영화 sum"))   # ko
print(detect_language("A dark dramatic movie 포스터"))  # en
```
