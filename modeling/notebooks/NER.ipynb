```python
from gliner import GLiNER
from transformers import AutoTokenizer, AutoModelForTokenClassification, pipeline

class NER:
    def __init__(self):
        self.gliner = GLiNER.from_pretrained("taeminlee/gliner_ko")
        self.tta_labels = [
            "ARTIFACTS",      # 인공물(작품명, 영화제목 등)
            "ANIMAL",         # 동물
            "CIVILIZATION",   # 문명, 문화
            "DATE",           # 날짜, 연도 등
            "EVENT",          # 사건, 행사
            "STUDY_FIELD",    # 연구/학문 분야
            "LOCATION",       # 장소, 지역, 위치
            "MATERIAL",       # 재료, 소재
            "ORGANIZATION",   # 조직, 단체, 회사
            "PERSON",         # 사람(인물)
            "PLANT",          # 식물
            "QUANTITY",       # 수량, 수치
            "TIME",           # 시간, 시각
            "TERM",           # 용어
            "THEORY",         # 이론
        ]
        tokenizer = AutoTokenizer.from_pretrained("Leo97/KoELECTRA-small-v3-modu-ner")
        model = AutoModelForTokenClassification.from_pretrained("Leo97/KoELECTRA-small-v3-modu-ner")
        self.leo97_ner = pipeline("token-classification", model=model, tokenizer=tokenizer, aggregation_strategy="simple")

        
    def predict(self, text:str, model:str) -> dict:
        match model:
            case "gliner":
                results = self.gliner.predict_entities(text, self.tta_labels)
            case "leo97":
                results = self.leo97_ner(text)

        return results
    
    def _postprocess(self, results:dict, model:str) -> dict:
        
```

```python
from transformers import AutoTokenizer, AutoModelForTokenClassification, pipeline

model_name = "Leo97/KoELECTRA-small-v3-modu-ner"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForTokenClassification.from_pretrained(model_name)

ner = pipeline("token-classification", model=model, tokenizer=tokenizer, aggregation_strategy="simple")

text = "피터 잭슨은 반지의 제왕 영화의 감독이다. 반지의 제왕은 로맨스다. 장르는 코미디다."
results = ner(text)
for ent in results:
    print(f"텍스트: {ent['word']}")
    # entity_group은 엔터티의 실제 태그(B-PERSON, B-LOCATION 등)들을 개념적으로 묶어준 그룹 이름입니다.
    # 예를 들어 태그(B-PERSON, I-PERSON)가 모두 'PERSON'이라는 entity_group으로 묶입니다.
    # 즉, 태그는 BIO 포맷의 세부 레이블이고, entity_group은 그 태그들이 속하는 엔터티의 범주입니다.
    print(f"  엔터티 그룹: {ent['entity_group']}")  
    print(f"  신뢰도 점수: {ent['score']:.2f}")
    print(f"  시작 위치: {ent.get('start', 'N/A')}")
    print(f"  끝 위치: {ent.get('end', 'N/A')}")
    print("-" * 30)
```

| 태그               | 의미        | 영화 도메인 예시           |

| `B-ARTIFACTS`    | 작품명, 영화제목 | “기생충”, “인터스텔라”      |
| `B-ORGANIZATION` | 제작사, 배급사  | “넷플릭스”, “CJ ENM”    |
| `B-LOCATION`     | 촬영지, 배경   | “부산”, “뉴질랜드”        |
| `B-EVENT`        | 영화제, 시상식  | “칸 영화제”, “아카데미 시상식” |
